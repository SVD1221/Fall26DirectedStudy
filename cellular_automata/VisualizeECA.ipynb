```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
file = open("eca_output.txt")
fileString = file.read().split("\n")
```

```python
rules = [int(x) for x in fileString[1].split()]
width = int(fileString[0])
iterations = len(fileString[2])//(2*width)
numRules = len(rules)
shape = numRules, iterations, width
rules
```

```python
ECAs = np.zeros(shape)
for i in range(numRules):
  ECAs[i, :, :] = np.array([int(x) for x in fileString[i+2].split()]).reshape((-1, width))
```

```python
fig, ax = plt.subplots(3, 3, figsize=[10,10])
for i in range(numRules):
  ax[i//3, i % 3].set_title(f"{rules[i]}")
  ax[i//3, i % 3].imshow(ECAs[i])
  ax[i//3, i % 3].set_xticks([])
  ax[i//3, i % 3].set_yticks([])
for i in range(numRules, 9):
  ax[i//3, i % 3].remove()
```
