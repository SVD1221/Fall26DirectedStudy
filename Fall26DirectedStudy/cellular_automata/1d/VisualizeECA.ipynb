```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
```

```python
file = open("../output_files/1d/eca_output.txt")
fileString = file.read().split("\n")
```

```python
rules = [int(x) for x in fileString[1].split()]
width = int(fileString[0])
iterations = len(fileString[2])//(2*width)
numRules = len(rules)
shape = numRules, iterations, width
rules
```

```python
ECAs = np.zeros(shape)
for i in range(numRules):
  ECAs[i, :, :] = np.array([int(x) for x in fileString[i+2].split()]).reshape((-1, width))
```

```python
colors = ["#e74c3c", "#2ecc71", "#f1c40f", "#e67e22", "#3498db", "#9b59b6"]
cmap = ListedColormap(colors[3:5])
```

```python
fig, ax = plt.subplots(3, 3, figsize=[20,20])
plt.subplots_adjust(
    left=0.1,    # Left margin
    right=0.5,   # Right margin
    bottom=0.1,  # Bottom margin
    top=0.45,     # Top margin
    wspace=0.1,  # Width space between subplots
    hspace=0.0   # Height space between subplots
)
for i in range(numRules):
  ax[i//3, i % 3].set_title(f"{rules[i]}")
  ax[i//3, i % 3].imshow(ECAs[i], cmap=cmap)
  ax[i//3, i % 3].set_xticks([])
  ax[i//3, i % 3].set_yticks([])
for i in range(numRules, 9):
  ax[i//3, i % 3].remove()
```
